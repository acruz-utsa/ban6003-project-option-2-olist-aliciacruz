# Table Inventory

**Orders: 25,000 rows x 8 columns**
This table includes all orders from XX/XX/XXXX through XX/XX/XXXX. All of the fields in this table will be used in some form for the analysis. Order delivery timing could be an important factor in review score.

order_id: unique identifier associated with each order
customer_id: unique identifier associated with customer that placed the order
**order_status: [NEED VALUES]**
order_purchase_timestamp: date that the customer placed the order
**order_approved_at: date that the order was approved for shipping??**
**order_delivered_carrier_date: date that the carrier picked up the order??**
order_delivered_customer_date: date that the order was actually delivered to the customer
order_estimated_delivery_date: estimated order delivery date

**Customers: 25,000 rows x 5 columns**
This table includes a list of customers. This analysis will not use these fields. The only use of the location fields would be to determine whether there are delivery anomalies (faster or slower) associated with any of these locations, but delivery speed is already accounted for in the data contained in the Orders table. In doing a regression, if these location fields were included it would likely cause autocorrelation issues.

customer_id: unique identifier associated with the customer.
**customer_unique_id: ???**
customer_zip_code_prefix: the zip code associated with the customer's delivery address.
customer_city: the city associated with the customer's delivery address.
customer_state: the state associated with the customer's delivery address.

**Items: 28,343 rows x 7 columns**
This table contains one row per item associated with each order. It contains information on the items purchased and the seller associated with each item.

order_id: unique ID associated with each order, used to join with Orders table.
order_item_id: sequential number of the item in the order itself. For example, if there were 3 items in the order, the rows would be numbered 1, 2 and 3.
product_id: unique ID associated with the product ordered in each order line item.
seller_id: unique ID associated with the seller for the associated product.
**shipping_limit_date: ???**
price: price paid for each order line item.
freight_value: value of each order line item in terms of freight cost.

**Payments: 26,154 rows x 5 columns**
This table contains the payments made on each order. Each row represents a payment associated with the order. The analysis will use the installment data to determine whether using a payment plan impacted the review score.

order_id: unique ID associated with each order, used to join with Orders table.
payment_sequential: if the customer elected to pay in installments, this field indicates the sequence number of the payment.
payment_type: what method of payment was used.
payment_installments: how many payment installments are required to pay off the order.
payment_value: the currency value of each payment.

**Reviews: 24,934 rows x 7 columns**
This table contains the reviews that were left by customers on their orders. This analysis will use the review score field as the outcome variable.

review_id: unique identifier associated with each review.
order_id: unique ID associated with each order, used to join with orders table.
**review_score: score on scale of [X-X] that the customer provided on the order**
review_comment_title: title of the review comment
review_comment_message: text of the review message
review_creation_date: date that the review was created by the customer
review_answer_timestamp: date that the review was responded to by the seller

**Products: 13,344 rows x 9 columns**
This table contains product information for all of the products that the subject company sells. This analysis will use the product ID and product category fields to investigate whether the products ordered had an impact on reviews.

product_id: unique identifier associated with each product.
**product_category_name: [need values] the broader category that each product fits into**
product_name_lenght: number of characters in product name
product_description_lenght: number of characters in product description
product_photos_qty: number of photos of product
product_weight_g: product weight
product_length_cm: product length
product_height_cm: product height
product_width_cm: product width

**Sellers: 2,234 rows x 4 columns**
This table contains the list of sellers associated with the products. The analysis will use the seller_id field to determine if there is any correlation between the seller and the review score.

seller_id: unique identifier associated with the seller
seller_zip_code_prefix: seller zip code
seller_city: seller city
seller_state: seller state

**Geolocation_zip_prefix: 10,009 rows x 6 columns**
The analysis will not use any of these fields. It is unknown which geolocations this table references.

geolocation_zip_code_prefix
geolocation_lat
geolocation_lng
geolocation_city
geolocation_state
geolocation_records

**Translations: 71 rows x 2 columns**
This table contains translations of the product categories into English. The analysis will use the English versions of the category names.

product_category_name: name of the product category in Portuguese
product_category_name_english: translated name fo the product category in English.
